```python
## Explore the geographical distribution of the restaurants and identify the cities with the maximum and minimum number of restaurants.

city_counts = df["City"].value_counts()
city_counts
```

```python
print("City with maximum restaurants:", city_counts.idxmax())
print("Maximum number of restaurants:", city_counts.max())

print("City with minimum restaurants:", city_counts.idxmin())
print("Minimum number of restaurants:", city_counts.min())
```

```python
top_10_cities = city_counts.head(10)

plt.figure(figsize=(10, 6))
top_10_cities.plot(kind="bar")
plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
## Restaurant franchising explore the franchise with most national presence

franchise_presence = df.groupby("Restaurant Name")["City"].nunique().sort_values(ascending=False)

franchise_presence.head(10)
```

```python
## most nationally present franchise

print("Restaurant/Franchise with highest national presence:")
print(franchise_presence.idxmax())

print("Number of cities:")
print(franchise_presence.max())
```

```python
top_franchises = franchise_presence.head(10)

plt.figure(figsize=(10, 6))
top_franchises.plot(kind="bar")
plt.title("Top 10 Restaurant Franchises by City Presence")
plt.xlabel("Restaurant / Franchise")
plt.ylabel("Number of Cities")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
## Find the ratio between restaurants that allow table booking vs those that do not.

## Ratio of restaurants that allow table booking vs. those that do not

table_booking_counts = df["Has Table booking"].value_counts()

print(table_booking_counts)

table_booking_ratio = (
    df["Has Table booking"].value_counts(normalize=True) * 100
).round(2)

print("\nPercentage distribution:")
print(table_booking_ratio)
```

```python
plt.figure(figsize=(7, 5))

table_booking_counts.plot(kind="bar")

plt.title("Restaurants: Table Booking Availability")
plt.xlabel("Table Booking")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
## Find the percentage of restaurants providing online delivery.

## Percentage of restaurants providing online delivery

online_delivery_counts = df["Has Online delivery"].value_counts()

print(online_delivery_counts)

online_delivery_percentage = (
    df["Has Online delivery"].value_counts(normalize=True) * 100
).round(2)

print("\nPercentage distribution:")
print(online_delivery_percentage)
```

```python
plt.figure(figsize=(7, 5))

online_delivery_counts.plot(kind="bar")

plt.title("Restaurants: Online Delivery Availability")
plt.xlabel("Online Delivery")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Calculate the difference in number of votes between restaurants that deliver online and restaurants that do not.

## Difference in votes between restaurants that deliver and those that do not

delivery_votes = df.groupby("Has Online delivery")["Votes"].sum()

print("Total votes by online delivery status:")
print(delivery_votes)

difference_votes = abs(
    delivery_votes["Yes"] - delivery_votes["No"]
)

print("\nDifference in total votes:", difference_votes)
```

```python
plt.figure(figsize=(7, 5))

delivery_votes.plot(kind="bar")

plt.title("Total Votes: Online Delivery vs No Online Delivery")
plt.xlabel("Online Delivery")
plt.ylabel("Total Votes")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

## PART 3 — CUISINE ANALYSIS

```python
## Find the top 10 cuisines served across cities.

## Top 10 cuisines served across cities

cuisine_data = df["Cuisines"].dropna().str.split(", ").explode()

top_10_cuisines = cuisine_data.value_counts().head(10)

print("Top 10 cuisines:")
print(top_10_cuisines)
```

```python
plt.figure(figsize=(10, 6))

top_10_cuisines.plot(kind="bar")

plt.title("Top 10 Cuisines Served")
plt.xlabel("Cuisine")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
## What is the maximum and minimum number of cuisines that a restaurant serves? Also, 
## which is the most served cuisine across the restaurant for each city?

## Maximum & minimum cuisines per restaurant
## Maximum and minimum number of cuisines served by a restaurant

cuisine_count_per_restaurant = (
    df["Cuisines"]
    .dropna()
    .apply(lambda x: len(x.split(",")))
)

print("Minimum number of cuisines served:",
      cuisine_count_per_restaurant.min())

print("Maximum number of cuisines served:",
      cuisine_count_per_restaurant.max())

print("\nDistribution of number of cuisines:")
print(cuisine_count_per_restaurant.value_counts().sort_index())
```

```python
## Most served cuisine in each city

city_cuisine = (
    df[["City", "Cuisines"]]
    .dropna()
    .assign(Cuisines=lambda x: x["Cuisines"].str.split(", "))
    .explode("Cuisines")
)

most_served_cuisine_by_city = (
    city_cuisine.groupby(["City", "Cuisines"])
    .size()
    .reset_index(name="Count")
    .sort_values(["City", "Count"], ascending=[True, False])
    .drop_duplicates("City")
)

most_served_cuisine_by_city.head(20)
```

```python
## How ratings are distributed among the various factors?

rating_counts = df["Aggregate rating"].value_counts().sort_index()

print("Rating Distribution:")
print(rating_counts)
```

```python
plt.figure(figsize=(10, 6))

rating_counts.plot(kind="bar")

plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
## Rating vs Average Cost

## Relationship between rating and average cost

rating_cost = df.groupby("Aggregate rating")["Average Cost for two"].mean()

print("Average Cost for Two by Rating:")
print(rating_cost.round(2))
```

```python
plt.figure(figsize=(10, 6))

rating_cost.plot(kind="line", marker="o")

plt.title("Average Cost for Two vs Restaurant Rating")
plt.xlabel("Aggregate Rating")
plt.ylabel("Average Cost for Two")

plt.tight_layout()
plt.show()
```

```python
## Rating vs Votes

## Relationship between rating and votes

rating_votes = df.groupby("Aggregate rating")["Votes"].mean()

print("Average Votes by Rating:")
print(rating_votes.round(2))
```

```python
plt.figure(figsize=(10, 6))

rating_votes.plot(kind="line", marker="o")

plt.title("Average Votes vs Restaurant Rating")
plt.xlabel("Aggregate Rating")
plt.ylabel("Average Votes")

plt.tight_layout()
plt.show()
```

```python
## Rating vs Online Delivery

## Relationship between rating and online delivery

rating_delivery = df.groupby("Has Online delivery")["Aggregate rating"].mean()

print("Average Rating by Online Delivery:")
print(rating_delivery.round(2))
```

```python
plt.figure(figsize=(7, 5))

rating_delivery.plot(kind="bar")

plt.title("Average Restaurant Rating: Online Delivery vs No Online Delivery")
plt.xlabel("Online Delivery")
plt.ylabel("Average Aggregate Rating")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
## Rating vs Table Booking

## Relationship between rating and table booking

rating_booking = df.groupby("Has Table booking")["Aggregate rating"].mean()

print("Average Rating by Table Booking:")
print(rating_booking.round(2))
```

```python
plt.figure(figsize=(7, 5))

rating_booking.plot(kind="bar")

plt.title("Average Restaurant Rating: Table Booking vs No Table Booking")
plt.xlabel("Table Booking")
plt.ylabel("Average Aggregate Rating")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

## Conclusion
The analysis shows that restaurants offering online delivery and table booking generally have higher average ratings than restaurants without these facilities. Restaurants with online delivery have an average rating of 3.25 compared with 2.46 for those without online delivery. Similarly, restaurants with table booking have an average rating of 3.44 compared with 2.56 for restaurants without table booking.

This suggests that additional customer convenience services such as online delivery and table booking may be associated with better restaurant ratings.

```python
# Most served cuisine across restaurants for each city

cuisine_city = df[['City', 'Cuisines', 'Restaurant ID']].copy()

# Remove missing cuisine values
cuisine_city = cuisine_city.dropna(subset=['City', 'Cuisines'])

# Split multiple cuisines into separate rows
cuisine_city['Cuisines'] = cuisine_city['Cuisines'].str.split(',')

# Create one row for each cuisine
cuisine_city = cuisine_city.explode('Cuisines')

# Remove extra spaces
cuisine_city['Cuisines'] = cuisine_city['Cuisines'].str.strip()

# Count restaurants for each cuisine in each city
cuisine_count = (
    cuisine_city
    .groupby(['City', 'Cuisines'])['Restaurant ID']
    .nunique()
    .reset_index(name='Restaurant Count')
)

# Find the most served cuisine for each city
most_served_cuisine = cuisine_count[
    cuisine_count.groupby('City')['Restaurant Count'].transform('max')
    == cuisine_count['Restaurant Count']
]

# Sort by city
most_served_cuisine = most_served_cuisine.sort_values(['City', 'Restaurant Count'], ascending=[True, False])

most_served_cuisine.head(20)
```

```python
# Save most served cuisine by city for Tableau

most_served_cuisine.to_csv("most_served_cuisine_by_city.csv", index=False)

print("File saved successfully!")
```
